```
import pandas as pd
import numpy as np
import plotly.express as px
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow.keras import layers, models

# 3D візуал залежності
df = pd.read_csv('/content/alarms.csv')
fig = px.scatter_3d(df, x='sound', y='distance', z='alarm')
fig.show()

# Підготовка датафрейму
x = df[['sound', 'distance', 'visibility']].values
y = df['alarm'].values

# Розділяємо дані на тренувальний та тестовий набори,
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Ось тут ми запускаємо нашу "качалку" нейронної мережі
model = models.Sequential([
    layers.Dense(64, activation='relu', input_shape=(X_train.shape[1],)),
    layers.Dense(64, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Тренуємо модель.
model.fit(X_train, y_train,
          epochs=125,
          batch_size=32,
          validation_data=(X_test, y_test))

# Оцінка моделі: ну ось ми і готові, тестуємо, що вийшло
loss, accuracy = model.evaluate(X_test, y_test)
print(f"Тест завершено. Втрата: {loss}, Точність: {accuracy}")  # Ось тут ми виводимо результат

# Прогнозування — це як передбачити майбутнє, тільки по нашим даним
new_data = np.array([
    [0.7, 0.1, 0.7],
    [0.3, 0.7, 0.3],
    [0.4, 0.5, 0.3],
    [0.6, 0.5, 0.7],
    [0.9, 0.1, 0.9]
])

# Ось і наші прогнози — кожна точка виведе "0" або "1", чи є насправді сигнал тривоги
predictions = model.predict(new_data)
print("Прогнози:")  # Ось що буде далі
print(predictions)  # Ось і результат!
```

```
import pandas as pd
import numpy as np
import plotly.express as px
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import tensorflow as tf
from tensorflow.keras import layers, models

# Завантаження даних
df = pd.read_csv('/content/top-1000-trending-youtube-videos.csv')
fig = px.scatter_3d(df, x='Video views', y='Likes', z='Dislikes', color="Category")
fig.show()
# Очищення: залишаємо потрібні колонки
df = df[['Video views', 'Likes', 'Dislikes', 'published']].dropna()

# Конвертація чисел: видаляємо коми та перетворюємо в int
for col in ['Video views', 'Likes', 'Dislikes']:
    df[col] = df[col].str.replace(',', '', regex=False).astype(np.int64)

# Цільова змінна: вірусність (більше 100 млн переглядів)
df['viral'] = df['Video views'].apply(lambda x: 1 if x >= 100_000_000 else 0)

# Підготовка X та y
X = df[['Likes', 'Dislikes', 'published']].values
y = df['viral'].values

# Масштабування ознак
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Розділення на тренувальні та тестові
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Побудова моделі
model = models.Sequential([
    layers.Dense(64, activation='relu', input_shape=(X_train.shape[1],)),
    layers.Dense(64, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])

# Компіляція
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Навчання
model.fit(X_train, y_train, epochs=100, batch_size=32, validation_data=(X_test, y_test))

# Оцінка
loss, accuracy = model.evaluate(X_test, y_test)
print(f'Loss: {loss:.4f}, Accuracy: {accuracy:.4f}')

# Прогнози
new_samples = np.array([
    [2000000, 100000, 2022],
    [15000, 500, 2017]
])

new_samples_scaled = scaler.transform(new_samples)
predictions = model.predict(new_samples_scaled)
print("Ймовірність вірусності (у відсотках):")
for i, prob in enumerate(predictions):
    percent = prob[0] * 100
    print(f"Відео {i + 1}: {percent:.4f}%")
```
